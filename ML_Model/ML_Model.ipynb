```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import mean_absolute_error, r2_score
import matplotlib.pyplot as plt
import seaborn as sns

# --- 1. Data Loading from Delhi_NCR_1990_2022_Safdarjung.csv ---
# Load your actual Delhi rainfall data.
try:
    # Attempt to load the CSV file.
    # The date column is 'time' and the rainfall column is 'prcp' based on inspection.
    data = pd.read_csv(
        'Delhi_NCR_1990_2022_Safdarjung.csv',
        parse_dates=['time'], # Parse the 'time' column as datetime objects
        index_col='time'      # Set 'time' column as the DataFrame index
    )
    print("CSV loaded successfully. First 5 rows:")
    print(data.head())

    # --- ENSURE INDEX IS DATETIMEINDEX ---
    # This is the crucial fix for the AttributeError
    if not isinstance(data.index, pd.DatetimeIndex):
        print("Converting index to DatetimeIndex...")
        data.index = pd.to_datetime(data.index, errors='coerce') # Coerce errors to NaT
        
        # --- FIX: Drop rows where date parsing failed by filtering the index ---
        initial_rows = len(data)
        data = data[data.index.notna()] # Filter out rows where the index is NaT
        if len(data) < initial_rows:
            print(f"Dropped {initial_rows - len(data)} rows due to unparseable dates in 'time' column (NaT in index).")
        print(f"Index type after conversion: {type(data.index)}")


    # --- Data Cleaning and Preparation for the loaded CSV ---
    # Rename 'prcp' to 'Rainfall' for consistency
    if 'prcp' in data.columns:
        data.rename(columns={'prcp': 'Rainfall'}, inplace=True)
        print("Renamed 'prcp' to 'Rainfall'.")
    elif 'Rainfall' not in data.columns: # Fallback if 'prcp' is also missing
        raise ValueError("Rainfall column ('prcp') not found. Please ensure your CSV has a 'prcp' or 'Rainfall' column, or update the code to match your column name.")

    # Convert Rainfall to numeric, coercing errors to NaN
    data['Rainfall'] = pd.to_numeric(data['Rainfall'], errors='coerce')

    # Handle missing values: Fill NaNs in 'Rainfall' with 0 (assuming no rainfall if missing)
    data['Rainfall'].fillna(0, inplace=True)

    # Check for other common weather columns and handle their potential NaNs
    # Based on the CSV, we have tavg, tmin, tmax. Let's map them to a general 'Temperature' concept.
    potential_features = {
        'TAVG': ['tavg', 'TAVG', 'Temp_Avg'], # Average Temperature
        'TMIN': ['tmin', 'TMIN', 'Temp_Min'], # Minimum Temperature
        'TMAX': ['tmax', 'TMAX', 'Temp_Max'], # Maximum Temperature
        # No humidity or pressure columns observed in the provided snippet, so keeping them commented
        # 'Humidity': ['RHUM', 'Humidity'],
        # 'Pressure': ['PRES', 'Pressure'],
        # 'Wind_Speed': ['WIND', 'WindSpeed']
    }

    actual_features_found = []
    for feature_name, col_options in potential_features.items():
        for col_option in col_options:
            if col_option in data.columns:
                # Rename if necessary to a consistent name for feature use (e.g., 'TAVG' to 'TAVG')
                if col_option != feature_name:
                     data.rename(columns={col_option: feature_name}, inplace=True)
                     print(f"Renamed '{col_option}' to '{feature_name}'.")

                data[feature_name] = pd.to_numeric(data[feature_name], errors='coerce')
                # Impute with mean, only if the column was found and has NaNs
                if data[feature_name].isnull().any():
                     data[feature_name].fillna(data[feature_name].mean(), inplace=True)
                actual_features_found.append(feature_name)
                break # Move to next feature_name once a match is found
        # If no match is found for a potential feature, it's simply not added to actual_features_found
        # and a warning is printed later if it's explicitly sought after.


    print("\n--- Data Description (after loading and initial cleaning) ---")
    print(data.describe())
    print(f"\nTotal data points (days): {len(data)}")

except FileNotFoundError:
    print("Error: 'Delhi_NCR_1990_2022_Safdarjung.csv' not found.")
    print("Please ensure the CSV file is in the same directory as this notebook, or provide the full path.")
    print("Using simulated data for demonstration purposes.")
    # Fallback to simulated data if file not found
    np.random.seed(42)
    num_days = 50 * 365
    dates = pd.date_range(start='1975-01-01', periods=num_days, freq='D')
    temperature = (25 + 10 * np.sin(dates.dayofyear * (2 * np.pi / 365)) + 5 * np.random.randn(num_days)).clip(10, 45)
    humidity = (70 + 20 * np.sin(dates.dayofyear * (2 * np.pi / 365) + np.pi/2) + 10 * np.random.randn(num_days)).clip(30, 100)
    pressure = (1000 + 5 * np.sin(dates.dayofyear * (2 * np.pi / 365) + np.pi) + 2 * np.random.randn(num_days)).clip(980, 1030)
    wind_speed = (10 + 5 * np.random.randn(num_days)).clip(0, 30)

    rainfall = np.zeros(num_days)
    for i, date in enumerate(dates):
        base_rain = 0.5 * np.random.rand()
        if 6 <= date.month <= 9:
            base_rain = (5 + 15 * np.random.rand())
            base_rain += (humidity[i] / 100) * 10
            if date.month in [7, 8]:
                base_rain += 5 * np.random.rand()
        if np.random.rand() > 0.8:
            rainfall[i] = max(0, base_rain + np.random.normal(0, 3))
        else:
            rainfall[i] = 0
    rainfall[rainfall < 0] = 0
    rainfall = rainfall.clip(0, 200)

    data = pd.DataFrame({
        'TAVG': temperature, # Renamed to TAVG for consistency with actual CSV
        'Humidity': humidity,
        'Pressure': pressure,
        'Wind_Speed': wind_speed,
        'Rainfall': rainfall
    }, index=dates)
    actual_features_found = ['TAVG', 'Humidity', 'Pressure', 'Wind_Speed'] # Update for simulated too
    print("\n--- Using Simulated Data (first 5 rows) ---")
    print(data.head())


# Ensure the index is sorted by date for time-series operations
data.sort_index(inplace=True)

# --- 2. Feature Engineering (Crucial for time series) ---
# Add lagged rainfall as a feature. This tells the model about recent rainfall.
# You can add more lags (e.g., 7 days, 30 days) and other meteorological features
# from previous days.
data['Rainfall_Lag1'] = data['Rainfall'].shift(1)
data['Rainfall_Lag7'] = data['Rainfall'].shift(7) # Rainfall 7 days ago
data['Avg_Rain_Last_7_Days'] = data['Rainfall'].rolling(window=7).mean().shift(1)

# Add time-based features (month, day of year, etc.)
# These operations now rely on the index being a DatetimeIndex, which is now ensured
data['Month'] = data.index.month
data['DayOfYear'] = data.index.dayofyear
data['Year'] = data.index.year # Useful for understanding trends over years

# Drop rows with NaN values created by shifting (first few rows)
data.dropna(inplace=True)

print("\n--- Data after Feature Engineering (first 5 rows) ---")
print(data.head())
print(f"Total data points after dropping NaNs: {len(data)}")


# --- 3. Data Scaling ---
# Scale features to a range suitable for neural networks (e.g., 0 to 1)

# Dynamically build the features list based on what's available in the data
# Starting with lagged rainfall and time features
features = ['Rainfall_Lag1', 'Rainfall_Lag7', 'Avg_Rain_Last_7_Days', 'Month', 'DayOfYear', 'Year']
# Add temperature features if they were found in the data
for feat in actual_features_found:
    if feat in data.columns and feat not in features: # Avoid duplicates, add if present
        features.append(feat)

target = 'Rainfall'

print(f"\nFeatures used for training: {features}")


# Create feature matrix X and target vector y
X = data[features].values
y = data[target].values.reshape(-1, 1) # Reshape for scaler

scaler_X = MinMaxScaler(feature_range=(0, 1))
scaler_y = MinMaxScaler(feature_range=(0, 1)) # Scale target as well

X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y)

# Combine scaled features and target for sequence creation
scaled_data = np.concatenate((X_scaled, y_scaled), axis=1)

# --- 4. Prepare Data for LSTM (Sequence Creation) ---
# LSTMs need data in the format [samples, timesteps, features]
# We'll use a sequence of `look_back` days to predict rainfall for the next day.
look_back = 30 # Use the last 30 days of data to predict rainfall for the next day

def create_sequences(dataset, look_back):
    X, y = [], []
    # Ensure there are enough data points to create at least one sequence
    if len(dataset) <= look_back:
        print(f"Warning: Not enough data points ({len(dataset)}) to create sequences with look_back={look_back}. Adjust look_back or provide more data.")
        return np.array([]), np.array([]) # Return empty arrays
    
    for i in range(len(dataset) - look_back):
        a = dataset[i:(i + look_back), :-1] # Features for 'look_back' days
        y.append(dataset[i + look_back, -1]) # Target (Rainfall) for the next day
        X.append(a)
    return np.array(X), np.array(y)

X_sequences, y_sequences = create_sequences(scaled_data, look_back)

if X_sequences.shape[0] == 0:
    print("Error: No sequences created. Please check data size and look_back period.")
    # For a Jupyter environment, it's better to provide clear messages and let the user debug
    raise RuntimeError("Cannot proceed: No sequences created. Please check data size and look_back period.")


print(f"\nShape of X_sequences (samples, timesteps, features): {X_sequences.shape}")
print(f"Shape of y_sequences (samples,): {y_sequences.shape}")

# Split into training and testing sets (time-series split)
# It's crucial to maintain the temporal order in time series splitting.
train_size = int(len(X_sequences) * 0.8)
# Ensure train_size is at least 1 and test_size is at least 1 for valid split
if train_size == 0 or len(X_sequences) - train_size == 0:
    print("Error: Not enough data after sequence creation to split into training and testing sets.")
    print("Please ensure you have enough data points (after dropping NaNs) for look_back and split.")
    raise RuntimeError("Cannot proceed: Not enough data for train/test split.")

X_train, X_test = X_sequences[0:train_size], X_sequences[train_size:len(X_sequences)]
y_train, y_test = y_sequences[0:train_size], y_sequences[train_size:len(y_sequences)]

print(f"Training sequences: {len(X_train)}")
print(f"Testing sequences: {len(X_test)}")

# --- 5. Build the LSTM Model ---
model = Sequential()
# Add LSTM layers
model.add(LSTM(units=50, return_sequences=True, input_shape=(look_back, X_sequences.shape[2])))
model.add(Dropout(0.2)) # Dropout for regularization
model.add(LSTM(units=50, return_sequences=False)) # Last LSTM layer doesn't return sequences
model.add(Dropout(0.2))
model.add(Dense(units=1)) # Output layer for predicting a single rainfall value

model.compile(optimizer='adam', loss='mean_squared_error')
model.summary()

# --- 6. Train the Model ---
# Early stopping to prevent overfitting
early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)

print("\n--- Training the LSTM model... ---")
history = model.fit(X_train, y_train, epochs=50, batch_size=32,
                    validation_split=0.1, callbacks=[early_stopping], verbose=1)
print("Model training complete!")

# Plot training history
plt.figure(figsize=(10, 6))
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Model Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# --- 7. Make Predictions and Inverse Transform ---
train_predict_scaled = model.predict(X_train)
test_predict_scaled = model.predict(X_test)

# Inverse transform to get actual rainfall values
train_predict = scaler_y.inverse_transform(train_predict_scaled)
test_predict = scaler_y.inverse_transform(test_predict_scaled)

y_train_actual = scaler_y.inverse_transform(y_train.reshape(-1, 1))
y_test_actual = scaler_y.inverse_transform(y_test.reshape(-1, 1))

# --- 8. Model Evaluation ---
mae_train = mean_absolute_error(y_train_actual, train_predict)
r2_train = r2_score(y_train_actual, train_predict)

mae_test = mean_absolute_error(y_test_actual, test_predict)
r2_test = r2_score(y_test_actual, test_predict)

print(f"\n--- Model Evaluation (Training Set) ---")
print(f"Mean Absolute Error (MAE): {mae_train:.2f} mm")
print(f"R-squared (R2) Score: {r2_train:.2f}")

print(f"\n--- Model Evaluation (Testing Set) ---")
print(f"Mean Absolute Error (MAE): {mae_test:.2f} mm")
print(f"R-squared (R2) Score: {r2_test:.2f}")

# --- 9. Visualization of Predictions ---
# Plot actual vs. predicted for the test set
plt.figure(figsize=(14, 7))
plt.plot(y_test_actual, label='Actual Rainfall', alpha=0.7)
plt.plot(test_predict, label='Predicted Rainfall', alpha=0.7, linestyle='--')
plt.title('Actual vs. Predicted Rainfall (Test Set)')
plt.xlabel('Time Steps (Days)')
plt.ylabel('Rainfall (mm)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# You might want to zoom in on a smaller section for better visualization
plt.figure(figsize=(14, 7))
plt.plot(y_test_actual[200:400], label='Actual Rainfall', alpha=0.8)
plt.plot(test_predict[200:400], label='Predicted Rainfall', alpha=0.8, linestyle='--')
plt.title('Actual vs. Predicted Rainfall (Zoomed In Test Set)')
plt.xlabel('Time Steps (Days)')
plt.ylabel('Rainfall (mm)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()


# --- 10. Example of making a new prediction ---
print("\n--- Example New Prediction ---")
# To make a real future prediction, you would need the last 'look_back'
# days of actual, most recent data for all features, including lagged rainfall.
# Then you would scale this data and pass it to the model.

# Ensure there's enough data for a new prediction
if len(data) < look_back:
    print(f"Cannot make a new prediction: Data length ({len(data)}) is less than look_back ({look_back}).")
else:
    # Get the last `look_back` rows of data for the features
    last_days_data_raw = data[features].tail(look_back).values
    last_days_data_scaled = scaler_X.transform(last_days_data_raw)

    # Reshape for LSTM input (1 sample, look_back timesteps, number of features)
    new_prediction_input = last_days_data_scaled.reshape(1, look_back, X_sequences.shape[2])

    # Make prediction
    predicted_rainfall_scaled_new = model.predict(new_prediction_input)

    # Inverse transform to get actual rainfall value
    predicted_rainfall_new = scaler_y.inverse_transform(predicted_rainfall_scaled_new)

    print(f"Predicted rainfall for the next day (based on last {look_back} days of data): {predicted_rainfall_new[0][0]:.2f} mm")
```
